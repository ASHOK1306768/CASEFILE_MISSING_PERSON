# Module 4: Movement Pattern Analysis & Spatial Clustering (K-Means & DBSCAN)


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import sys
sys.path.append("../src")
from preprocessing import clean_gps_data
from clustering import fit_kmeans, fit_dbscan, get_cluster_centroids, save_model

df_traj = pd.read_csv("../data/processed/indian_trajectories.csv")
df_clean = clean_gps_data(df_traj)

# 1. K-Means Clustering
kmeans_model, km_labels = fit_kmeans(df_clean, n_clusters=8)
df_clean["Location_Cluster"] = km_labels

centroids = get_cluster_centroids(df_clean, "Location_Cluster")
print("=== Cluster Centroids (Stay Point Hotspots) ===")
print(centroids)

# 2. DBSCAN Clustering
db_model, db_labels = fit_dbscan(df_clean, eps_km=0.5, min_samples=5)
print("\nDBSCAN Discovered Clusters:", len(set(db_labels)) - (1 if -1 in db_labels else 0))

# Save Model
save_model(kmeans_model, "../models/clustering_model.pkl")
print("Saved models/clustering_model.pkl")

```
