```python
import pandas as pd
import numpy as np
import sys
sys.path.append("../src")
from prediction import train_location_model, save_model

df_cases = pd.read_csv("../data/synthetic/synthetic_case_data.csv")

features = ["Age_Group", "Gender", "Weather", "Day", "Average_Distance", "Average_Speed", "Time_Since_Last_Seen", "Usual_Area", "Previous_Area"]

# Train Random Forest Classifier
rf_model, metrics, feat_imp = train_location_model(df_cases, features, target="Target_Area", model_type="random_forest")

print("=== Model Evaluation Metrics (PDF Section 11) ===")
for k, v in metrics.items():
    if k != "Confusion_Matrix":
        print(f"{k}: {v:.4f}")

print("\n=== Explainable AI Feature Importances (PDF Module 9) ===")
print(feat_imp)

save_model(rf_model, "../models/location_model.pkl")
print("\nSaved models/location_model.pkl")

```
