# Module 3 & 4: Exploratory Data Analysis & Movement Profiles


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import sys
sys.path.append("../src")
from preprocessing import clean_gps_data
from feature_engineering import add_movement_features, user_summary

df_traj = pd.read_csv("../data/processed/indian_trajectories.csv")
df_clean = clean_gps_data(df_traj)
df_feat = add_movement_features(df_clean)

summary = user_summary(df_feat)
print("=== User Movement Summary ===")
print(summary.head(10))

plt.figure(figsize=(10, 4))
plt.hist(df_feat["Speed_KMH"].dropna(), bins=30, color="teal", edgecolor="black")
plt.title("Distribution of GPS Movement Speed (KMH) - Indian Metros")
plt.xlabel("Speed (km/h)")
plt.ylabel("Frequency")
plt.grid(True, linestyle="--", alpha=0.5)
print("Exploratory Data Analysis Complete!")
```
